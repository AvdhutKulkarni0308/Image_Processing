```python
import cv2
import matplotlib.pyplot as plt
from google.colab import files
import numpy as np
```

```python
# Step 2: Upload an image
uploaded = files.upload()
```

```python
# Step 3: Read the uploaded image
image_path = next(iter(uploaded))
image = cv2.imread(image_path, cv2.IMREAD_GRAYSCALE)
```

```python
# Step 4: Apply Gaussian Blur to reduce noise
blurred_image = cv2.GaussianBlur(image, (5, 5), 1.4)
```

```python
# Step 5: Apply Sobel edge detector
edges = cv2.Sobel(blurred_image, cv2.CV_64F, 1, 0, ksize=3)
```

```python
# Step 6: Display the original image and the edge-detected image
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Edge Detected Image(Sobel)')
plt.imshow(edges, cmap='gray')
plt.axis('off')
plt.show()
```

```python
# step 7: Apply Canny edge detector
edges = cv2.Canny(blurred_image, 7, 21)
```

```python
# Step 6: Display the original image and the edge-detected image
plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Original Image')
plt.imshow(image, cmap='gray')
plt.axis('off')

plt.subplot(1, 2, 2)
plt.title('Edge Detected Image(Canny)')
plt.imshow(edges, cmap='gray')
plt.axis('off')
plt.show()
```
